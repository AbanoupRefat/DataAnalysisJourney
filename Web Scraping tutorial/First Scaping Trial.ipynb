```python
from bs4 import BeautifulSoup
import lxml
import requests
```

```python
quotes = []
authors = []

base_url =  'https://quotes.toscrape.com'
def get_html(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.content, 'lxml')
    return soup
```

```python
def Extract_quote_and_author(soup):
# Extracting quotes and authors from the web page using BeautifulSoup
    for quote in soup.find_all('div',{'class':'quote'}):
        quote_text = quote.find('span',{'class':'text'}).text
        author_name = quote.find('small',{'class':'author'}).text
        quotes.append(quote_text)
        authors.append(author_name)

# extract url of next page
def next_page(soup):
    next_li = soup.find('li', {'class': 'next'})    
    if next_li:
        next_a = next_li.find('a')  # Find the <a> tag within <li>
        if next_a:
            return base_url + next_a.get('href')  # Construct the full URL
    return None
```

```python

# Loop for up to 5 pages
url = base_url
for i in range(5):
    if not url:  # Stop if there are no more pages
        break

    # Get HTML and extract data
    soup = get_html(url)
    Extract_quote_and_author(soup)

    # Move to the next page
    url = next_page(soup)

    print(f"Page {i + 1} scraped.")

```

```python
import pandas as pd

df = pd.DataFrame({'Quote': quotes, 'Author': authors})

print(df)
df.to_csv('new_quotes.csv',index=False)
```
